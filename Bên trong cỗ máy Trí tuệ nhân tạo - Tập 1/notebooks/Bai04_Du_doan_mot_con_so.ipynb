## Chặng 1: Chạy
### Ô 1. Sổ bán hàng 20 ngày của quán trà sữa cạnh trường

```python
import numpy as np

nhiet_do = np.array([24, 25, 26, 26, 27, 28, 28, 29, 30, 30,
                     31, 31, 32, 32, 33, 33, 34, 35, 35, 36])   # độ C lúc trưa
so_ly = np.array([50, 58, 60, 65, 64, 72, 69, 77, 78, 83,
                  82, 88, 87, 92, 95, 91, 100, 101, 106, 107])  # số ly bán ra

print("Số ngày:", len(nhiet_do))
print("Nhiệt độ trung bình:", nhiet_do.mean(), "| Số ly trung bình:", so_ly.mean())
```

### Ô 2. Kiểm tra phép tính tay ở mục 5.5 (5 ngày)

```python
x = np.array([26, 28, 30, 32, 34])
y = np.array([60, 73, 78, 91, 98])

dx = x - x.mean()            # độ lệch của x so với trung bình
dy = y - y.mean()            # độ lệch của y so với trung bình
a = (dx * dy).sum() / (dx ** 2).sum()
b = y.mean() - a * x.mean()
print("a =", round(a, 2), "| b =", round(b, 2))
print("Dự đoán cho 31 độ:", round(a * 31 + b, 1), "ly")
```

### Ô 3. Để thư viện làm thay: khớp đường thẳng cho cả 20 ngày

```python
from sklearn.linear_model import LinearRegression

X = nhiet_do.reshape(-1, 1)       # sklearn cần X là bảng: 20 hàng, 1 cột
mo_hinh = LinearRegression()
mo_hinh.fit(X, so_ly)             # HUẤN LUYỆN: tìm a và b tốt nhất

a, b = mo_hinh.coef_[0], mo_hinh.intercept_
print(f"Đường thẳng: số ly = {a:.2f} x nhiệt độ + ({b:.2f})")
print("Dự đoán cho 31 độ:", mo_hinh.predict([[31]]).round(1))
```

### Ô 4. Mô hình sai trung bình bao nhiêu ly?

```python
from sklearn.metrics import mean_absolute_error, r2_score

du_doan = mo_hinh.predict(X)
print("5 ngày đầu - thật   :", so_ly[:5])
print("5 ngày đầu - dự đoán:", du_doan[:5].round(1))
mae = mean_absolute_error(so_ly, du_doan)
print("Sai số tuyệt đối trung bình (MAE):", round(mae, 2), "ly")
print("Hệ số R2:", round(r2_score(so_ly, du_doan), 3))
```

### Ô 5. Bắt mô hình đoán ở nơi nó chưa từng thấy

```python
for t in [10, 30, 45]:
    print(f"{t} độ: mô hình đoán {mo_hinh.predict([[t]])[0]:.0f} ly")
```

### Ô 6. Một ngày bất thường làm lệch cả đường thẳng
Ngày hội trường: trời 30 độ nhưng quán bán tới 300 ly.

```python
X_moi = np.append(nhiet_do, 30).reshape(-1, 1)
y_moi = np.append(so_ly, 300)
mh2 = LinearRegression().fit(X_moi, y_moi)
print(f"Trước: số ly = {a:.2f} x nhiệt độ + ({b:.2f})")
print(f"Sau  : số ly = {mh2.coef_[0]:.2f} x nhiệt độ + ({mh2.intercept_:.2f})")
print("Dự đoán cho 24 độ - trước:", round(a * 24 + b),
      "| sau:", round(mh2.predict([[24]])[0]))
```

### Ô 7. Hai thứ cùng tăng chưa chắc cái này gây ra cái kia
Số liệu dưới đây được **dựng lên để minh hoạ**: cả số kem bán ra và số người đi bơi đều phụ thuộc vào nhiệt độ.

```python
rng = np.random.default_rng(0)
nhiet = rng.uniform(22, 37, 60)                      # 60 ngày
kem = 8 * nhiet - 120 + rng.normal(0, 12, 60)        # phụ thuộc nhiệt độ
di_boi = 5 * nhiet - 90 + rng.normal(0, 10, 60)      # cũng phụ thuộc nhiệt độ

r = np.corrcoef(kem, di_boi)[0, 1]
print("Hệ số tương quan giữa kem và người đi bơi:", round(r, 2))
```

### Ô 8. Vẽ hình

```python
import matplotlib.pyplot as plt
plt.scatter(nhiet_do, so_ly, label="20 ngày thật")
plt.plot(nhiet_do, mo_hinh.predict(X), color="orange", label="Đường thẳng của mô hình")
plt.xlabel("Nhiệt độ lúc trưa (độ C)"); plt.ylabel("Số ly bán ra"); plt.legend(); plt.show()
```
