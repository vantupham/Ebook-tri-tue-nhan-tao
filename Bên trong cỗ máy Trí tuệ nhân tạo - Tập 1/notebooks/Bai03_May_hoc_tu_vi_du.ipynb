## Chặng 1: Chạy
### Ô 1. Dữ liệu: 24 tin nhắn đã gắn nhãn

```python
# 12 tin rác và 12 tin thường (tin mẫu do nhóm tác giả tự soạn)
rac = [
    "chúc mừng bạn trúng thưởng nhận quà miễn phí",
    "nhận quà miễn phí bấm vào link ngay",
    "vay tiền nhanh lãi thấp bấm link nhận ngay",
    "tài khoản của bạn bị khóa bấm link để xác minh",
    "bạn trúng thưởng xe máy liên hệ ngay để nhận",
    "khuyến mãi sốc nạp thẻ nhận quà miễn phí",
    "việc nhẹ lương cao làm tại nhà liên hệ ngay",
    "nhận ngay mã giảm giá bấm vào link",
    "cơ hội trúng thưởng điện thoại miễn phí hôm nay",
    "vay tiền online duyệt nhanh không cần thế chấp",
    "bạn có bưu phẩm chưa nhận bấm link để xác minh",
    "nạp thẻ hôm nay nhận khuyến mãi gấp đôi",
]
thuong = [
    "mai nhớ mang vở toán nhé",
    "tối nay học nhóm bài toán không",
    "mẹ nhận tiền chưa con",
    "chiều nay đá bóng ở sân trường nhé",
    "cô dặn mai kiểm tra mười lăm phút môn lý",
    "con về muộn mẹ để phần cơm nhé",
    "cuối tuần đi xem phim với lớp không",
    "bạn gửi mình bài tập tiếng anh với",
    "hôm nay trời mưa to nhớ mang áo mưa",
    "sinh nhật An tối thứ bảy bạn đi không",
    "mình đến cổng trường rồi bạn ra chưa",
    "bố chuyển tiền học thêm cho con rồi nhé",
]
tin = rac + thuong
nhan = ['rác'] * len(rac) + ['thường'] * len(thuong)
print('Số tin:', len(tin))
```

### Ô 2. Chia dữ liệu: phần để học, phần để kiểm tra

```python
from sklearn.model_selection import train_test_split

# Giữ lại 25% số tin làm 'đề thi' mà máy chưa từng thấy
tin_hoc, tin_thi, nhan_hoc, nhan_thi = train_test_split(
    tin, nhan, test_size=0.25, random_state=0, stratify=nhan)
print('Học:', len(tin_hoc), 'tin | Thi:', len(tin_thi), 'tin')
```

### Ô 3. Biến chữ thành số (đếm từ) và huấn luyện

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

dem = CountVectorizer()               # máy đếm từ
X_hoc = dem.fit_transform(tin_hoc)    # mỗi tin thành một hàng số
mo_hinh = MultinomialNB()             # thuật toán học: Naive Bayes
mo_hinh.fit(X_hoc, nhan_hoc)          # HUẤN LUYỆN
print('Máy đã học', X_hoc.shape[1], 'từ khác nhau')
```

### Ô 4. Chấm điểm trên 'đề thi'

```python
diem = mo_hinh.score(dem.transform(tin_thi), nhan_thi)
print('Đoán đúng', round(diem * 100), '% số tin trong đề thi')
```

### Ô 5. Thử với tin nhắn mới

```python
tin_moi = [
    'chúc mừng bạn nhận quà miễn phí',
    'tối nay học toán không',
    'mẹ chuyển tiền cho con rồi bấm link xem nhé',
    'ch.úc m.ừng b.ạn tr.úng th.ưởng',
]
X_moi = dem.transform(tin_moi)
for t, d, p in zip(tin_moi, mo_hinh.predict(X_moi), mo_hinh.predict_proba(X_moi)):
    print(f'{d:7} | chắc là rác: {p[0]:.0%} | {t}')
```

## Chặng 3: Sửa
Xem máy đã học được gì: từ nào 'nghiêng' về tin rác nhất?

```python
import numpy as np
tu = dem.get_feature_names_out()
# hiệu số càng lớn thì từ càng nghiêng về 'rác'
nghieng = mo_hinh.feature_log_prob_[0] - mo_hinh.feature_log_prob_[1]
thu_tu = np.argsort(nghieng)
print('Nghiêng về RÁC nhất   :', list(tu[thu_tu[-8:]][::-1]))
print('Nghiêng về THƯỜNG nhất:', list(tu[thu_tu[:8]]))
```

### Kiểm tra phép tính tay ở mục 5.5
Sách tính tay ra 95,7% cho tin *bấm link nhận tiền*. Thư viện có ra đúng như vậy không?

```python
tin6 = ["nhận quà miễn phí bấm link", "trúng thưởng nhận quà ngay",
        "vay tiền nhanh nhận tiền bấm link", "mẹ nhận tiền chưa con",
        "mai nhớ mang vở toán", "tối nay học nhóm toán không"]
nhan6 = ["rác"] * 3 + ["thường"] * 3
dem6 = CountVectorizer()
mh6 = MultinomialNB(alpha=1.0)        # alpha=1.0 là "cộng thêm 1"
mh6.fit(dem6.fit_transform(tin6), nhan6)
print(mh6.predict_proba(dem6.transform(["bấm link nhận tiền"])).round(3))
```

**Việc của em:** quay lại Ô 5, sửa danh sách `tin_moi` thành 4 tin do em tự nghĩ ra. Hãy cố viết **một tin rác mà máy tưởng là tin thường**. Em đã làm thế nào?

## Chặng 4: Tự làm
1. Thêm 5 tin rác và 5 tin thường của riêng em vào Ô 1 (xoá tên người, số điện thoại thật), chạy lại từ đầu. Điểm ở Ô 4 thay đổi thế nào?
2. Đổi thuật toán học sang cây quyết định bằng ô dưới đây và so sánh.

```python
from sklearn.tree import DecisionTreeClassifier

cay = DecisionTreeClassifier(random_state=0)
cay.fit(X_hoc, nhan_hoc)
print('Cây quyết định đoán đúng', round(cay.score(dem.transform(tin_thi), nhan_thi) * 100), '%')
```
