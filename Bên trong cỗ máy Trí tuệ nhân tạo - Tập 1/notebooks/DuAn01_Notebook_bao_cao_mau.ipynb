## Phần 4. Khám phá bằng số và biểu đồ

```python
import matplotlib.pyplot as plt

# TODO: chọn hai cột số để vẽ
plt.scatter(sach["cot_x"], sach["cot_y"])
plt.xlabel("cot_x"); plt.ylabel("cot_y")
plt.title("TODO: tiêu đề nói lên điều cần thấy")
plt.show()
```

## Phần 5. Một mô hình đơn giản
Chọn **một** trong ba: hồi quy (Bài 4), phân loại (Bài 3), gom cụm (Bài 6). Nhớ chia tập học và tập kiểm tra nếu có nhãn (Bài 5).

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

X = sach[["cot_x"]]          # TODO
y = sach["cot_y"]            # TODO
X_hoc, X_thi, y_hoc, y_thi = train_test_split(X, y, test_size=0.3, random_state=0)
mo_hinh = LinearRegression().fit(X_hoc, y_hoc)
print("Hệ số a:", mo_hinh.coef_[0].round(2), "| MAE trên tập thi:",
      round(mean_absolute_error(y_thi, mo_hinh.predict(X_thi)), 2))
```

## Phần 6. Kiểm tra công bằng (Bài 7)
Nếu dữ liệu chia được thành nhóm (khối lớp, cách đi học...), so kết quả của mô hình giữa các nhóm.

```python
# TODO: thay "nhom" bằng cột chia nhóm của nhóm em
sach["sai_so"] = (sach["cot_y"] - mo_hinh.predict(sach[["cot_x"]])).abs()
print(sach.groupby("nhom")["sai_so"].mean().round(2))
```
