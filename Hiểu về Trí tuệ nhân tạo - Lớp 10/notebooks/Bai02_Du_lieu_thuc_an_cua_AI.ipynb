## Chặng 3: Sửa
### Ô 6. Làm sạch từng bước

```python
sach = df.copy()                       # luôn giữ lại bản gốc

# 1) Bỏ hàng bị nhập trùng
sach = sach.drop_duplicates()

# 2) Chuẩn hoá chữ: bỏ dấu cách thừa, đổi về chữ thường
sach["di_hoc_bang"] = sach["di_hoc_bang"].str.strip().str.lower()

# 3) Giờ ngủ lớn hơn 24 là vô lý: đánh dấu là "không biết"
sach.loc[sach["gio_ngu"] > 24, "gio_ngu"] = None

# 4) Điền ô trống bằng trung vị của cột
for cot in ["gio_ngu", "diem_toan"]:
    sach[cot] = sach[cot].fillna(sach[cot].median())

print("Kích thước:", sach.shape,
      "| Số ô trống:", int(sach.isna().sum().sum()))
print("Giờ ngủ - trung bình:", round(sach["gio_ngu"].mean(), 2),
      "| trung vị:", sach["gio_ngu"].median())
print(sach["di_hoc_bang"].value_counts())
```

### Ô 7. Hỏi dữ liệu một câu
Các bạn ngủ từ 7,5 giờ trở lên có điểm Toán khác các bạn ngủ ít hơn không?

```python
ngu_du = sach[sach["gio_ngu"] >= 7.5]
ngu_it = sach[sach["gio_ngu"] < 7.5]
print(f"Ngủ từ 7,5 giờ: {len(ngu_du)} bạn, điểm Toán trung bình {ngu_du['diem_toan'].mean():.2f}")
print(f"Ngủ dưới 7,5 giờ: {len(ngu_it)} bạn, điểm Toán trung bình {ngu_it['diem_toan'].mean():.2f}")
```
