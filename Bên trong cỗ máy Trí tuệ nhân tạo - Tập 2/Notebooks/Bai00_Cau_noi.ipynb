```python
import numpy as np

a = np.array([1, 2, 3])
M = np.arange(6).reshape(2, 3)          # bảng 2 hàng, 3 cột
print("a * 2   =", a * 2)               # nhân từng phần tử
print("a @ a   =", a @ a)               # tích vô hướng: 1 + 4 + 9
print("M.shape =", M.shape, "| M.T.shape =", M.T.shape)
print("M @ a   =", M @ a)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

tin = ["trúng thưởng iphone bấm link ngay", "khuyến mãi lớn nạp thẻ nhận quà",
       "bấm link nhận quà miễn phí", "trúng thưởng tiền mặt gọi ngay",
       "mai lớp mình kiểm tra toán", "chiều nay đá bóng ở sân trường",
       "nhớ mang sách văn nhé", "cô dặn nộp bài tập lý"]
nhan = ["rác"] * 4 + ["thường"] * 4
dem = CountVectorizer()
loc = MultinomialNB().fit(dem.fit_transform(tin), nhan)
moi = ["bấm link nhận iphone", "nhớ kiểm tra bài tập"]
for cau, du_doan in zip(moi, loc.predict(dem.transform(moi))):
    print(f"{du_doan:7} <- {cau}")
```

```python
nhiet_do = np.array([26, 28, 30, 32, 34])
so_ly = np.array([61, 71, 80, 89, 99])
a, b = np.polyfit(nhiet_do, so_ly, 1)     # đường thẳng khớp nhất (Bài 4)
print(f"số ly = {a:.1f} × nhiệt độ + ({b:.0f})")
```

```python
w, b = np.array([4.7]), -61               # cùng đường thẳng, viết kiểu Tập 2
x = np.array([31])
print("w · x + b =", round(float(w @ x + b), 1))   # nhân rồi cộng: một nơ-ron (Bài 9)
```
